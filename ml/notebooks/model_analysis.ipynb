# AVATAR ML — model incelemesi

Bu notebook `ml/train.py` çıktılarını okur. Veri `synthetic_calibrated`;
sonuçlar gerçek saha doğrulaması değildir.

```python
import json
from pathlib import Path

import pandas as pd

REPORTS = Path('../reports')
PROCESSED = Path('../data/processed')
comparison = json.loads((REPORTS / 'model_comparison.json').read_text(encoding='utf-8'))
list(comparison)
```

```python
rows = []
for group, payload in comparison.items():
    for target, report in payload['metadata']['targets'].items():
        selected = report['selected']
        rows.append({
            'group': group,
            'target': target,
            'model': report['selected_model'],
            'cv_mae': selected['cv_mae_mean'],
            'test_mae': selected['test_mae'],
            'test_r2': selected['test_r2'],
            'beats_baseline': report['beats_baseline'],
        })
pd.DataFrame(rows).sort_values(['group', 'target'])
```

```python
from IPython.display import Image, display

for name in [
    'yield_yield_pct_actual_vs_predicted.png',
    'butchery_lamb_total_saleable_weight_actual_vs_predicted.png',
    'butchery_cattle_total_saleable_weight_actual_vs_predicted.png',
]:
    display(Image(filename=str(REPORTS / 'plots' / name)))
```
